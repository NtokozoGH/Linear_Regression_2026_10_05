```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
file_path='retail_sales_dataset.csv'
df = pd.read_csv(file_path)
df
```

```python
#checking information in the csv file 
df.info()
```

```python
#checking number of rows and columns 
df.shape
```

```python
#transaction ID 
mean_id = df['Transaction ID'].mean()
median_id = df['Transaction ID'].median()
mode_id = df['Transaction ID'].mode()[0]
```

```python
#display for transaction id
print(f"Mean of Transaction ID: {mean_id}")
print(f"Median of Transaction ID: {median_id}")
print(f"Mode of Transaction ID: {mode_id}")
```

```python
# 2. Convert the 'Date' column to a proper datetime format
df['Date'] = pd.to_datetime(df['Date'])
```

```python
# Date looking at three aspects 
mean_date = df['Date'].mean()
median_date = df['Date'].median()
mode_date = df['Date'].mode()
```

```python
#display results for date 
print(f"--- Date Analysis ---")
print(f"Mean (Average Date):   {mean_date.strftime('%Y-%m-%d')}")
print(f"Median (Middle Date):  {median_date.strftime('%Y-%m-%d')}")
print(f"Mode (Top Sales Date): {mode_date.dt.strftime('%Y-%m-%d').values[0]}")
```

```python
#Extract numbers and convert to numeric format
numeric_customer_id = pd.to_numeric(df['Customer ID'].str.extract(r'(\d+)', expand=False), errors='coerce')
```

```python
#Customer ID 
mean_cust = numeric_customer_id.mean()
median_cust = numeric_customer_id.median()
mode_cust = numeric_customer_id.mode()
```

```python
print("--- Customer ID Numeric Values ---")
print(f"Mean:   {mean_cust}")
print(f"Median: {median_cust}")
print(f"Mode:   {mode_cust.values}")
```

```python
# 1. Calculate the Mode directly from the text column
mode_gender = df['Gender'].mode()[0]
```

```python
# 2. Convert Gender to numbers (Female = 0, Male = 1) to find mean and median
numeric_gender = df['Gender'].map({'Female': 0, 'Male': 1})

mean_gender = numeric_gender.mean()
median_gender = numeric_gender.median()

# Convert the median number back into a readable label
median_gender_label = 'Male' if median_gender == 1 else 'Female' if median_gender == 0 else 'Exact 50/50 Split'

# Display the results
print("--- Gender Statistical Values ---")
print(f"Mean (As a numeric score): {mean_gender:.3f}")
print(f"Median (Middle Value):     {median_gender_label} ({median_gender})")
print(f"Mode (Most Frequent):      {mode_gender}")
```

```python
#get age 
mean_age = df["Age"].mean()
median_age = df["Age"].median()
mode_ages = df["Age"].mode().tolist()
```

```python
print(f"Mean Age:    {mean_age:.2f}")
print(f"Median Age:  {median_age}")
print(f"Mode Age(s): {mode_ages}")
```

```python
#Product Category
# 1. Get the Mode (Most frequent category)
most_frequent_categories = df["Product Category"].mode().tolist()
```

```python
category_counts = df["Product Category"].value_counts()
category_percentages = df["Product Category"].value_counts(normalize=True) * 100

# Print the results neatly
print(f"Most Popular Category (Mode): {', '.join(most_frequent_categories)}")
print("\n--- Full Breakdown ---")
for category, count in category_counts.items():
    pct = category_percentages[category]
    print(f"{category}: {count} items ({pct:.1f}%)")
```

```python
#Quantity
mean_qty = df["Quantity"].mean()
median_qty = df["Quantity"].median()
mode_qty = df["Quantity"].mode().tolist()
```

```python
print(f"Mean Quantity:    {mean_qty:.2f}")
print(f"Median Quantity:  {median_qty}")
print(f"Mode Quantity(s): {mode_qty}")
```

```python
#Price per Unit 
mean_price = df["Price per Unit"].mean()
median_price = df["Price per Unit"].median()
mode_prices = df["Price per Unit"].mode().tolist()
```

```python
print(f"Mean Price per Unit:    ${mean_price:.2f}")
print(f"Median Price per Unit:  ${median_price:.2f}")
print(f"Mode Price per Unit(s): {', '.join([f'${p:.2f}' for p in mode_prices])}")
```

```python
#total amount 
mean_total = df["Total Amount"].mean()
median_total = df["Total Amount"].median()
mode_totals = df["Total Amount"].mode().tolist()
```

```python
print(f"Mean Total Amount:    ${mean_total:.2f}")
print(f"Median Total Amount:  ${median_total:.2f}")
print(f"Mode Total Amount(s): {', '.join([f'${t:.2f}' for t in mode_totals])}")
```

```python
def Stats():

    print("Average Total Amount:", df['Total Amount'].mean())

    print("Median:", df['Total Amount'].median())

    print("Mode:", df['Total Amount'].mode())

    print('-------------------------------------')

    print("Average Price per Unit:", df['Price per Unit'].mean())

    print("Median:", df['Price per Unit'].median())

    print("Mode:", df['Price per Unit'].mode())

    print('-------------------------------------')

    print("Average Quantity:", df['Quantity'].mean())

    print("Median:", df['Quantity'].median())

    print("Mode:", df['Quantity'].mode())
```

```python
Stats()
```

```python
df.groupby('Gender')['Product Category'].value_counts().unstack().plot(kind='bar')
```

```python
#import matplotlib.pyplot as plt

# 1. Create the plot and assign it to a variable (ax)
ax = df.groupby('Gender')['Product Category'].value_counts().unstack().plot(kind='bar')

# 2. Loop through the bars and add the labels
for container in ax.containers:
    ax.bar_label(container, fmt='%d', padding=3)

# 3. Display the final plot
plt.show()
```
